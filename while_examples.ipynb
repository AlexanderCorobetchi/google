# Python while loop examples

This notebook shows several examples of the `while` loop in Python: a basic counter loop, using `break` and `continue`, the `while`-`else` clause, and a safe input-based loop.

```python
# Example 1: Basic counter loop - prints numbers 1 to 5
i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
# Example 2: Using break - stop when i == 3
i = 1
while i <= 5:
    if i == 3:
        print('Stopping at', i)
        break
    print(i)
    i += 1
```

```python
# Example 3: Using continue - skip printing even numbers
i = 0
while i < 6:
    i += 1
    if i % 2 == 0:
        continue
    print(i)
# prints 1,3,5
```

```python
# Example 4: while-else - else runs if loop wasn't terminated by break
i = 1
while i <= 3:
    print(i)
    i += 1
else:
    print('Loop finished without break')
```

```python
# New program: read one character and print its Unicode code and whether it's ASCII
ch = input('Введите символ: ')
# Take only the first character if the user entered more
if len(ch) == 0:
    print('Ничего не введено')
else:
    c = ch[0]
    code = ord(c)
    is_ascii = code < 128
    print()
    print('Код Unicode:', code)
    print('ASCII символ:', is_ascii)
```
